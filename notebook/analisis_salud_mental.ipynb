# Analisis de Salud Mental de Estudiantes Internacionales

## Resumen
Analisis exploratorio de 286 estudiantes de universidad japonesa (2018) - 201 internacionales. Objetivo: evaluar como la duracion de la estancia (stay) impacta en indicadores de salud mental.

## Dataset
- Fuente: Estudio de universidad internacional japonesa (2018)
- Registros: 286 totales, 201 internacionales filtrados
- Variables clave: inter_dom, stay, todep (depresion PHQ-9), tosc (conexion social SCS), toas (estres aculturacion ASISS)

## Hallazgos principales
- La cantidad de estudiantes internacionales disminuye con los años: 95 en año 1 vs 1 en año 10.
- La depresion PHQ-9 se mantiene entre 7-9 puntos en estancias de 1-4 años.
- El estres de aculturacion se mantiene alto (>77) incluso despues de 4+ años, indicando dificultad de integracion sostenida.


```python
import pandas as pd
students = pd.read_csv("data/students.csv")
students.head()
```

```python
SELECT
  stay,
  COUNT(*) AS count_int,
  ROUND(AVG(todep), 2) AS average_phq,
  ROUND(AVG(tosc), 2) AS average_scs,
  ROUND(AVG(toas), 2) AS average_as
FROM students
WHERE inter_dom = 'Inter'
GROUP BY stay
ORDER BY stay DESC;
```
